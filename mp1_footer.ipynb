## Reflection
In this mini project, combined with the lectures one, two, and three, I was able to conduct *almost* every single exercise clearly documented in both Jupyter Notebook and PDF format. Lecture one focused on setting and configuring a reproducible environment, where I extended it using a devcontainer for better developer user experience and a highly reproducible environment. 

In lecture two, I locked in a naive Mandelbrot Python runtime of $4.113s$ with NumPy of $0.578s$ resulting in a speedup of $6.85\times$. I also conducted a memory usage profiling between Python and NumPy, Python's usage slowly growing over time as compared to NumPy's single large memory allocation.

In lecture three, I confirmed my suspicions with line profiling that $99\%$ of my naive python implementation runtime was spent appending and computing a single point. I also implemented a compiled version using Numba, resulting in a runtime of $0.048s$, a speedup from naive Python of $86.6\times$. In the last lecture's extensions, I plotted the speedup of all implementations so far, with the parallel having the greatest speedup of about $560\times$.

| Implementation | Time   | Speedup |
| -------------- | -------| ------- |
| Naive          | 4.113s | N/A     |
| NumPy          | 0.578s | 6.85x   |
| Numba          | 0.048s | 86.6x   |
