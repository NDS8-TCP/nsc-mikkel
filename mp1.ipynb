### ****Milestone 2:** Full vectorized Mandelbrot (Target: 40 min)**

****Your Task:** Replace Python loops with NumPy operations**

****Naive approach has 3 nested loops:****
1. **Loop over rows (`i`)**
2. **Loop over columns (`j`) → selects pixel/point $c$**
3. **Loop over iterations (`n`) → computes $z = z^2 + c$ until escape**

**Hints: `np.zeros like(C)`, `np.zeros(C.shape, dtype=int)`**

#### ****NumPy approach — eliminate loops 1 & 2:****

- **Initialize `Z` and `M` arrays (same shape as `C`)**
- **Keep only loop 3 (iterations), operate on *all* pixels at once**
- **Boolean mask: `mask = np.abs(Z) <= 2`**
- **Update only unescaped points: `Z[mask] = Z[mask]**2 + C[mask]`**
- **Increment iteration count: `M[mask] += 1`**

````python
# From slide 29
import time
import statistics


def benchmark(func, *args, n_runs=3):
    """Time func, return median of n_runs."""
    times = []
    for _ in range(n_runs):
        t0 = time.perf_counter()
        result = func(*args)
        times.append(time.perf_counter() - t0)

    median_t = statistics.median(times)
    print(f"Median: {median_t:.4f}s"
          f"(min={min(times):.4f}, max={max(times):.4f})")
    return median_t, result
````

````python
from mandelbrot import compute_mandelbrot as naive_python
from mandelbrot import compute_mandelbrot_numpy as numpy_python

# Mostly copied from slide 29
t1, M1 = benchmark(naive_python, -2, 1, -1.5, 1.5, 1024, 1024, 100)
t2, M2 = benchmark(numpy_python, -2, 1, -1.5, 1.5, 1024, 1024, 100)

print(f"Naïve Python (median): {t1:.3f}s")
print(f"Numpy Python (median): {t2:.3f}s. Speedup of {t1/t2:.2f}x")
````

### ****Milestone 3:** Memory access pattern analysis (Target: 15 min)**

****Your Task:** Measure the effect of memory layout on performance**

**1. Create a large **square** array: `A = np.random.rand(N, N)` with $N = 10000$**

````python
N = 10_000
A = np.random.rand(N, N)
````

**2. Write a function that computes **row sums** by looping over rows:**
- **`for i in range(N): s = np.sum(A[i, :])`**

````python
def row_sum(A, N):
    for i in range(N):
        s = np.sum(A[i, :])
````

**3. Write a function that computes **column sums** by looping over columns:**
- **`for j in range(N): s = np.sum(A[:, j])`**

````python
def column_sum(A, N):
    for j in range(N):
        s = np.sum(A[:, j])
````

**4. Time both. Both loops run $N$ times — which is faster and why?**

````python
t_row, _ = benchmark(row_sum, A, N)
t_col, _ = benchmark(column_sum, A, N)

print(f"Row sum (median): {t_row:.3f}s")
print(f"Column sum (median): {t_col:.3f}s. Slowdown of {t_col/t_row:.2f}x")
````

**5. Now try with `A_f = np.asfortranarray(A)` (column-major). What changes?**

````python
A_f = np.asfortranarray(A)

t_row, _ = benchmark(row_sum, A_f, N)
t_col, _ = benchmark(column_sum, A_f, N)

print(f"Row sum (median): {t_row:.3f}s")
print(f"Column sum (median): {t_col:.3f}s. Speedup of {t_row/t_col:.2f}x")
````

### ****Milestone 3:** Implement Numba `@njit` and benchmark**

````python
from numba import njit


# From slide 36. Approach A (where I have filled in the gaps in the hybrid)
@njit
def mandelbrot_point_numba(c, max_iter=100):
    z = 0j
    for n in range(max_iter):
        if z.real*z.real+z.imag*z.imag > 4.0:
            return n
        z = z*z + c
    return max_iter


def mandelbrot_hybrid(xmin, xmax, ymin, ymax, width, height, max_iter=100):
    x = np.linspace(xmin, xmax, width)
    y = np.linspace(ymin, ymax, height)
    result = np.zeros((height, width), dtype=np.int32)

    # outer loops still in Python
    for i in range(height):
        for j in range(width):
            c = x[j] + 1j * y[i]
            result[i, j] = mandelbrot_point_numba(c, max_iter)
````

````python
# from slide 37. Approach B
from numba import njit
import numpy as np


@njit                       # <-- entire function compiled !
def mandelbrot_naive_numba(xmin, xmax, ymin, ymax, width, height, max_iter=100):
    """Fully JIT - compiled Mandelbrot --- structure identical to naive."""
    x = np.linspace(xmin, xmax, width)
    y = np.linspace(ymin, ymax, height)
    result = np.zeros((height, width), dtype=np.int32)

    for i in range(height):             # compiled loop
        for j in range(width):          # compiled loop
            c = x[j] + 1j * y[i]
            z = 0j                      # complex literal: type inference works!
            n = 0
            while n < max_iter and (z.real*z.real + z.imag*z.imag) <= 4.0:
                z = z*z + c
                n += 1
        result[i, j] = n
    return result


# Warm - up (triggers compilation --- don ’t time this!)
_ = mandelbrot_naive_numba(-2, 1, -1.5, 1.5, 64, 64)
````

#### ****Experiment 2:** Parallel Numba (advanced)**

````python
# From slide 46 with code filled in by me
from numba import njit, prange


@njit(parallel=True)
def mandelbrot_numba_parallel(xmin, xmax, ymin, ymax, width, height, max_iter=100):
    x = np.linspace(xmin, xmax, width)
    y = np.linspace(ymin, ymax, height)
    result = np.zeros((height, width), dtype=np.int32)

    for i in prange(height):    # Parallel!
        for j in range(width):
            c = x[j] + 1j * y[i]
            z = 0j
            n = 0
            while n < max_iter and (z.real*z.real + z.imag*z.imag) <= 4.0:
                z = z*z + c
                n += 1
        result[i, j] = n
    return result
````

- **Try `parallel=True` and `prange`**

````python
import numba as nb

t_parallel = bench(mandelbrot_numba_parallel,
                   -2, 1, -1.5, 1.5, 1024, 1024, 100, runs=100)

print(f"Using {nb.get_num_threads()} threads")
print(f"Numba parallel (median): {t_parallel:.3f}s")
````

- **Test $512^2$, $1024^2$, $4096^2$ (or larger)**

````python
grid_sizes = [512, 1024, 4096]
````

- **Plot grid size vs. time for all implementations**

````python
import json
from multiprocessing import Pool
from pathlib import Path
from mandelbrot import compute_mandelbrot as mandelbrot_naive
from mandelbrot import compute_mandelbrot_numpy as mandelbrot_numpy

implementations: list[callable] = [
    mandelbrot_naive,
    mandelbrot_numpy,
    mandelbrot_hybrid,
    mandelbrot_naive_numba,
    mandelbrot_numba_typed,
    mandelbrot_numba_typed_jit,
    # mandelbrot_numba_parallel # unsafe to compute in parallel in GNU OpenMP
]


def benchmark_implementation(task):
    print(task)
    impl, grid, runs = task
    time = bench(impl, -2, 1, -1.5, 1.5, grid, grid, 100, runs=runs)
    return (impl.__name__, grid, time)


def main(file: Path):
    benchmarking_tasks = [
        # func, grid, benchmark runs
        # 1 run or else this is going to take hours
        (impl, grid, 1) for grid in grid_sizes for impl in implementations
    ]
    with Pool() as p:
        results = p.map(benchmark_implementation, benchmarking_tasks)

    results_json = [{"impl": impl, "grid": grid, "time": time}
                    for impl, grid, time in results]
    # Also run mandelbrot_numba_parallel
    for grid in grid_sizes:
        time = bench(mandelbrot_numba_parallel, -2, 1, -
                     1.5, 1.5, grid, grid, 100, runs=1)
        results_json.append(
            {"impl": mandelbrot_numba_parallel.__name__, "grid": grid, "time": time})

    with file.open("w") as fp:
        json.dump(results_json, fp)


results_file = Path("l03_mandelbrot_set.json")
if not results_file.exists():
    print(f"{results_file} does not exists. Compute results... This will take multiple minutes!")
    main(results_file)
````

````python
with results_file.open("r") as fp:
    results = json.load(fp)

for impl in results:
    print(impl)
````

````python
# AI assisted
implementations: list[str] = sorted({entry["impl"] for entry in results})

for impl in implementations:
    # Filter entries for this implementation
    data: list[dict] = [entry for entry in results if entry["impl"] == impl]

    # Sort by grid size
    data = sorted(data, key=lambda x: x["grid"])

    grids = [entry["grid"] for entry in data]
    times = [entry["time"] for entry in data]

    plt.plot(grids, times, marker="o", label=impl)

plt.title("Grid size vs time for different implementations")
plt.xlabel("Grid width [pixels]")
plt.ylabel("Runtime [sec]")
plt.grid(alpha=0.5)
plt.legend()
plt.show()
````

- **Does speedup scale with problem size?**

````python
# AI assisted
# lookup: (impl, grid) -> time
lookup = {(r["impl"], r["grid"]): r["time"] for r in results}
grids = sorted({r["grid"] for r in results})

# Plot speedup
for impl in implementations:
    if impl == "compute_mandelbrot":
        continue  # Baseline speedup

    speedups = []
    valid_grids = []

    for grid in grids:
        t_base = lookup[("compute_mandelbrot", grid)]
        t_impl = lookup[(impl, grid)]
        speedups.append(t_base / t_impl)
        valid_grids.append(grid)

    plt.plot(valid_grids, speedups, marker="o", label=impl)

plt.title("Speedup compared to naïve Python")
plt.xlabel("Grid size width [pixels]")
plt.ylabel("Speedup [x]")
plt.grid(alpha=0.5)
plt.legend()
plt.show()
````

It appears so, but the parallel behaves differently with its speedups. I suspect that this is due to scheduling overhead, but still a massive speedup!
