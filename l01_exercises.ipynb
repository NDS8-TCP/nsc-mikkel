#### ****Step 4:** Measure execution time (10 min)**

- **Use `time.time()` or `timeit` to measure performance**

I have chosen to do both.

*Using `time.time()`*

````python
# Mostly copied from slide 38
import time

start = time.time()
image = compute_mandelbrot(-2, 1, -1.5, 1.5, 100, 100, 100)
end = time.time()
elapsed_time = end - start

print(f"Naïve Python (100x100): {elapsed_time:.3f}s")
````

*Using `timeit`*

````python
from timeit import timeit

elapsed_time = timeit(
    lambda: compute_mandelbrot(
        x_min=-2,
        x_max=1,
        y_min=-1.5,
        y_max=1.5,
        width=100,
        height=100,
        max_iter=100
    ),
    number=1
)

print(f"Naïve Python (100x100): {elapsed_time:.3f}s")
````
